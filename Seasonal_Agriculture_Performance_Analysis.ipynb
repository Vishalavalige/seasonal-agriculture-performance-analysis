## 6. Data Quality Analysis

Checking for missing values, duplicate records, and any inconsistencies before analysis.

```python
missing = df.isnull().sum()
missing = missing[missing > 0]
print("Columns with missing values:")
print(missing)
print("\nPercent missing:")
print((missing / len(df) * 100).round(2))
```

```python
print("Fully duplicated rows:", df.duplicated().sum())
print("Duplicate Farm_IDs:", df['Farm_ID'].duplicated().sum())
```

```python
# Handle missing values: impute within each season (season affects these values,
# so a season-wise median preserves seasonal patterns better than a global median)
for col in ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']:
    if df[col].isnull().sum() > 0:
        df[col] = df.groupby('Season')[col].transform(lambda x: x.fillna(x.median()))

print("Remaining missing values:", df.isnull().sum().sum())
```

## 10. Outlier Analysis

Investigating potential outliers using statistical and visual methods. We do not automatically delete outliers — we decide, based on evidence, whether an extreme value is a data-entry problem, a legitimate observation, or something requiring further investigation.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
sns.boxplot(x=df['Yield_Tonnes_Ha'], ax=axes[0])
axes[0].set_title('Yield_Tonnes_Ha - raw')
sns.boxplot(x=df['Profit_INR'], ax=axes[1])
axes[1].set_title('Profit_INR - raw')
plt.tight_layout()
plt.show()

print(df['Yield_Tonnes_Ha'].describe())
```

```python
# Cap extreme yield outliers using the IQR rule (winsorize rather than drop, so
# every farm record is still available for the seasonal comparisons later)
Q1, Q3 = df['Yield_Tonnes_Ha'].quantile([0.25, 0.75])
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR
outlier_count = (df['Yield_Tonnes_Ha'] > upper_bound).sum()
print(f"Yield values above upper bound ({upper_bound:.2f} t/ha): {outlier_count} ({outlier_count/len(df)*100:.1f}% of rows)")

df['Yield_Tonnes_Ha_capped'] = df['Yield_Tonnes_Ha'].clip(upper=upper_bound)
```

## 13. Seasonal Comparison

The core question of this project: does agricultural performance really differ by season, and is that difference statistically meaningful (not just random variation)?

```python
groups_yield = [g['Yield_Tonnes_Ha_capped'].values for _, g in df.groupby('Season')]
f_stat, p_val = stats.f_oneway(*groups_yield)
print(f"ANOVA - Yield across seasons: F = {f_stat:.3f}, p = {p_val:.6f}")

groups_profit = [g['Profit_INR'].values for _, g in df.groupby('Season')]
f_stat2, p_val2 = stats.f_oneway(*groups_profit)
print(f"ANOVA - Profit across seasons: F = {f_stat2:.3f}, p = {p_val2:.6f}")

print("\n=> Seasonal differences are statistically significant" if p_val < 0.05 and p_val2 < 0.05
      else "\n=> Not all differences are statistically significant")
```

```python
df['Is_Loss'] = df['Profit_INR'] < 0
loss_rate = df.groupby('Season')['Is_Loss'].mean().mul(100).round(1).reindex(['Kharif','Rabi','Zaid'])
print("Percentage of loss-making farms by season:")
print(loss_rate)

fig, ax = plt.subplots(figsize=(7,4.5))
loss_rate.plot(kind='bar', color=sns.color_palette('Set2'), ax=ax)
ax.set_title('Loss-Making Farms by Season (%)')
ax.set_ylabel('% of farms with negative profit')
plt.tight_layout()
plt.show()
```

## 15. Key Insights

- **Yield and profit both differ significantly by season** (ANOVA p < 0.05 for both) — this is not random noise.
- **Kharif is the strongest season** for both yield and profit; **Zaid is the weakest**, often running at a loss on average.
- **Loss-making farms increase steadily from Kharif to Zaid** — profitability, not just yield, is the central seasonal problem.
- **Some crops are far more season-sensitive than others** (see the Crop x Season heatmap) — certain crops keep relatively stable yield across seasons while others drop sharply outside their best season.
- **Regional patterns are not uniform** — some states follow the general seasonal trend closely, others diverge, suggesting local factors (soil, local practices) also matter.
- **Water efficiency and disease/pest risk both tend to move together with the profit pattern across seasons.**

## 16. Conclusion

Seasonal variation has a clear, statistically significant effect on agricultural performance in this dataset. Kharif consistently outperforms Rabi and Zaid on yield and profitability, while Zaid carries both the highest loss rate and the lowest water efficiency. These patterns are consistent enough to support practical, evidence-based recommendations:

1. Prioritize support (irrigation, credit, advisory) for farms in the Zaid season, where losses are most concentrated.
2. Match crop choice to its strongest season using the Crop x Season yield/profit heatmaps.
3. Promote higher-efficiency irrigation methods in seasons/regions where water efficiency is currently lowest.
4. Build season-specific disease/pest advisories ahead of higher-risk seasons.
5. Investigate state-level outliers separately, since not all regions follow the same seasonal pattern.
